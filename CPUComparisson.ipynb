```python
import time
import numpy as np
import cupy as cp

n = 10000

# CPU (NumPy, float32)
a = np.random.rand(n, n).astype(np.float32)
t = time.time()
np.dot(a, a)
print("CPU:", round(time.time() - t, 2), "s")

# GPU (CuPy, float32)
b = cp.random.rand(n, n, dtype=cp.float32)
cp.dot(b, b)                      # warm-up
cp.cuda.Device().synchronize()
t = time.time()
cp.dot(b, b)
cp.cuda.Device().synchronize()
print("GPU:", round(time.time() - t, 2), "s")
```
