```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

beta_1 = np.linspace(-2, 3, 51)
f_beta_1 = 4 * beta_1**2 - 10 * beta_1 + 25

plt.figure(figsize=(12, 7))
sns.lineplot(x=beta_1, y=f_beta_1, marker="o")
plt.axvline(x=1.25, color="mediumseagreen")
plt.axhline(y=18.75, color="mediumseagreen")
plt.xlabel("beta_1")
plt.ylabel("f(beta_1)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

beta_1 = np.linspace(-2, 8, 101)
g_beta_1 = beta_1**2 - 10 * beta_1 + 3 * np.abs(beta_1) + 25

plt.figure(figsize=(12, 7))
sns.lineplot(x=beta_1, y=g_beta_1, marker="o")
plt.axvline(x=3.5, color="mediumseagreen")
plt.axhline(y=12.75, color="mediumseagreen")
plt.xlabel("beta_1")
plt.ylabel("g(beta_1)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

beta_1 = np.linspace(-2, 8, 101)

error = (5 - beta_1)**2
ridge_penalty = 3 * beta_1**2
lasso_penalty = 3 * np.abs(beta_1)

fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)

sns.lineplot(x=beta_1, y=error, label="Prediction error", ax=axes[0])
sns.lineplot(x=beta_1, y=ridge_penalty, label="Penalty", ax=axes[0])
sns.lineplot(x=beta_1, y=error + ridge_penalty, label="Total", ax=axes[0])
axes[0].axvline(x=1.25, color="mediumseagreen")
axes[0].axhline(y=18.75, color="mediumseagreen")
axes[0].set(title="Ridge", xlabel="beta_1", ylabel="Value")
axes[0].grid(True)

sns.lineplot(x=beta_1, y=error, label="Prediction error", ax=axes[1])
sns.lineplot(x=beta_1, y=lasso_penalty, label="Penalty", ax=axes[1])
sns.lineplot(x=beta_1, y=error + lasso_penalty, label="Total", ax=axes[1])
axes[1].axvline(x=3.5, color="mediumseagreen")
axes[1].axhline(y=12.75, color="mediumseagreen")
axes[1].set(title="Lasso", xlabel="beta_1", ylabel="Value")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

beta_1 = np.linspace(-2, 8, 501)

error = (5 - beta_1)**2
ridge_penalty = 10 * beta_1**2
lasso_penalty = 10 * np.abs(beta_1)

fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)

sns.lineplot(x=beta_1, y=error, label="Prediction error", ax=axes[0])
sns.lineplot(x=beta_1, y=ridge_penalty, label="Penalty", ax=axes[0])
sns.lineplot(x=beta_1, y=error + ridge_penalty, label="Total", ax=axes[0])
axes[0].scatter(5/11, 250/11, color="red", s=80, zorder=5)
axes[0].axvline(x=5/11, color="red", linestyle="--")
axes[0].set(title="Ridge: chosen coefficient = 0.455",
            xlabel="beta_1", ylabel="Value")

sns.lineplot(x=beta_1, y=error, label="Prediction error", ax=axes[1])
sns.lineplot(x=beta_1, y=lasso_penalty, label="Penalty", ax=axes[1])
sns.lineplot(x=beta_1, y=error + lasso_penalty, label="Total", ax=axes[1])
axes[1].scatter(0, 25, color="red", s=80, zorder=5)
axes[1].axvline(x=0, color="red", linestyle="--")
axes[1].set(title="Lasso: chosen coefficient = 0",
            xlabel="beta_1", ylabel="Value")

for ax in axes:
    ax.set_xlim(-1, 2)
    ax.set_ylim(0, 50)
    ax.grid(True)

plt.tight_layout()
plt.show()
```
